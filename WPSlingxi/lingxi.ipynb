首先，WPS灵犀会用代码概览整个Excel文件，显示前20行数据以了解表格结构，WPS灵犀会读取print打印出来的内容

```python
import pandas as pd

# 读取Excel文件
file_path = '../ws_test_file_use.xlsx'
df = pd.read_excel(file_path, sheet_name='CONFIGURATION', header=None)

# 显示前20行数据以了解表格结构
print(df.head(20))
```

然后，WPS灵犀会确认表头的位置，提示词中提到了子表格标题为 "3G Configuration" ，WPS在前20行找到了该子表格的标题，灵犀会重新读取数据，以第14行作为表头。

```python

# 重新读取数据，跳过前13行，使用第14行作为表头
df = pd.read_excel(file_path, sheet_name='CONFIGURATION', header=14)

# 显示数据以确认结构
print("="*20)
print("显示数据以确认结构")
print(df.head(10))

# 提取所需的字段
# 根据表格结构，我们需要提取以下字段：
# SYSTEM MODULE, CELL, RF Module||Type, RF Module||QTY., UBBP type, BAND, TRX,
# ANTENNAS||NEW/SWAP/EXIST, ANTENNAS||Type, ANTENNAS||Qty.,
# RRU Cable||POWER LENGTH, RRU Cable||OPT LENGTH, RRU Cable||JUMPER TYPE,
# RRU Cable||JUMPER LENGTH(m), RRU Cable||Q-TY

# 创建提取的数据列表
extracted_data = []

# 添加表头
headers = [
    "SYSTEM MODULE", "CELL", "RF Module||Type", "RF Module||QTY.", "UBBP type", "BAND", "TRX",
    "ANTENNAS||NEW/SWAP/EXIST", "ANTENNAS||Type", "ANTENNAS||Qty.",
    "RRU Cable||POWER LENGTH", "RRU Cable||OPT LENGTH", "RRU Cable||JUMPER TYPE",
    "RRU Cable||JUMPER LENGTH(m)", "RRU Cable||Q-TY"
]
extracted_data.append(headers)

# 提取数据行（从第2行开始，因为第0行是表头，第1行是子表头）
for i in range(2, len(df)):
    row = df.iloc[i]
    # 检查是否有有效数据（CELL列不为空）
    if pd.notna(row.get('CELL', None)):
        extracted_row = [
            row.get('SYSTEM MODULE', ''),
            row.get('CELL', ''),
            row.get('RF MODULE', ''),
            row.get('QTY.', ''),
            row.get('UBBP type', ''),
            row.get('BAND', ''),
            row.get('TRX', ''),
            row.get('NEW/SWAP/EXIST', ''),
            row.get('Antenna Type', ''),
            row.get('Antenna Qty.', ''),
            row.get('POWER LENGTH(m)', ''),
            row.get('OPT LENGTH(m)', ''),
            row.get('JUMPTER TYPE', ''),
            row.get('JUMPTER LENGTH(m)', ''),
            row.get('Q-TY', '')
        ]
        extracted_data.append(extracted_row)

# 输出提取的数据
print("="*20)
print("输出提取的数据")
print(extracted_data)

```

为了方便起见，这里将输出提取的数据进行还原：

```python
df_view = pd.DataFrame(extracted_data[1:], columns=extracted_data[0])
print(df_view)
```
